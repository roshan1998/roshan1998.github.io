```python
import os
from pathlib import Path
from pprint import pprint

pdf_path = Path(r"C:\Users\roshan.patro\Downloads\Roshan-Kumar-Patro-Resume (1).pdf")
print(f"Resume PDF exists: {pdf_path.exists()}")
print(f"PDF path: {pdf_path}")

if pdf_path.exists():
    print(f"File size: {pdf_path.stat().st_size} bytes")
```

```python
try:
    from pypdf import PdfReader
    reader = PdfReader(str(pdf_path))
    print(f"Page count: {len(reader.pages)}")
    meta = reader.metadata
    print("Metadata:")
    pprint(meta)
    for i, page in enumerate(reader.pages[:3], start=1):
        text = page.extract_text() or ""
        print(f"\n--- Page {i} preview ---")
        print(text[:1500])
except Exception as e:
    print(f"Error reading PDF: {type(e).__name__}: {e}")
```

```python
# Load raw text from the PDF and inspect structure
try:
    from pypdf import PdfReader
    reader = PdfReader(str(pdf_path))
    pages = [page.extract_text() or "" for page in reader.pages]
    raw_text = "\n\n".join(pages)
    print(raw_text[:8000])
    print(f"\nTotal characters: {len(raw_text)}")
except Exception as e:
    print(f"Error extracting raw text: {type(e).__name__}: {e}")
```

```python
import re

contact_patterns = {
    'email': r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b',
    'phone': r'(?:\+?\d[\d\s().-]{7,}\d)',
    'linkedin': r'https?://(?:www\.)?linkedin\.com/in/[A-Za-z0-9\-_%]+/?',
    'github': r'https?://(?:www\.)?github\.com/[A-Za-z0-9_.\-]+/?',
}

for key, pattern in contact_patterns.items():
    matches = re.findall(pattern, raw_text, flags=re.IGNORECASE)
    print(f"{key}: {matches[:10]}")
```

```python
headline = []
for line in raw_text.splitlines():
    line = ' '.join(line.split())
    if len(line) > 20 and len(line) < 140:
        if any(token in line.lower() for token in ['software', 'engineer', 'developer', 'data', 'cloud', 'full stack', 'analyst']):
            headline.append(line)
print(headline[:15])
```

```python
import re

skill_keywords = [
    'python', 'java', 'javascript', 'typescript', 'c#', 'sql', 'aws', 'azure',
    'react', 'node.js', 'django', 'flask', 'spring', 'kubernetes', 'docker',
    'git', 'postgresql', 'mongodb', 'mysql', 'redis', 'power bi', 'tableau',
    'machine learning', 'data analysis', 'etl', 'api', 'microservices', 'html', 'css'
]

matches = []
for keyword in skill_keywords:
    if re.search(re.escape(keyword), raw_text, flags=re.IGNORECASE):
        matches.append(keyword)
print(matches)
```

```python
# Work experience parsing
experience_blocks = re.split(r'(?=\b(?:EXPERIENCE|WORK EXPERIENCE|EMPLOYMENT|Professional Experience)\b)', raw_text, flags=re.IGNORECASE)
print('experience blocks:', len(experience_blocks))
for block in experience_blocks[:3]:
    if len(block.strip()) > 50:
        print('\n--- block preview ---')
        print(block[:1200])
```

```python
education_blocks = re.split(r'(?=\b(?:EDUCATION|ACADEMIC|QUALIFICATIONS)\b)', raw_text, flags=re.IGNORECASE)
for block in education_blocks[:3]:
    if len(block.strip()) > 50:
        print('\n--- education preview ---')
        print(block[:1000])
```

```python
# Clean and normalize the extracted fields
import pandas as pd

normalized_skills = sorted(set(skill.lower().strip() for skill in matches if skill.strip()))
resume_df = pd.DataFrame({
    'section': ['contact', 'headline', 'skills'],
    'value': [
        {
            'email': re.findall(r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b', raw_text)[:1],
            'phone': re.findall(r'(?:\+?\d[\d\s().-]{7,}\d)', raw_text)[:1],
            'linkedin': re.findall(r'https?://(?:www\.)?linkedin\.com/in/[A-Za-z0-9\-_%]+/?', raw_text, flags=re.IGNORECASE)[:1],
        },
        headline[:5],
        normalized_skills,
    ]
})
print(resume_df)
```

```python
# Summary and export
print("Notebook workflow complete. Use the extracted data to personalize the resume website.")
resume_df.to_csv('resume_data.csv', index=False)
print('Saved structured data to resume_data.csv')
```
